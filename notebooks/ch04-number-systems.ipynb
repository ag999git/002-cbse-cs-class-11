## 1. Decimal to binary, octal and hex

`bin()`, `oct()` and `hex()` give the number with a prefix: **0b** for binary, **0o** for octal, **0x** for hex.

```python
n = 65
print(bin(n), oct(n), hex(n))
```

```python
print(int('1000001', 2), int('101', 8), int('41', 16), int('9C', 16))
```

## 3. The division ladder of the book (Road 1)

This small program divides again and again and reads the remainders upward, exactly as in section 4.3.

```python
def ladder(n, b):
    digits = '0123456789ABCDEF'
    remainders = []
    while n > 0:
        q, r = divmod(n, b)
        print(f'{n:>6} ÷ {b} = {q:<6} remainder {digits[r]}')
        remainders.append(digits[r])
        n = q
    print('Read upward:', ''.join(reversed(remainders)))

ladder(156, 8)
```

## 4. Place values (Road 2)

```python
def place_values(text, b):
    total = 0
    for position, digit in enumerate(reversed(text)):
        value = int(digit, 16)
        print(f'{digit} × {b}^{position} = {value * b ** position}')
        total += value * b ** position
    print('Total:', total)

place_values('3A5', 16)
```

## 5. Fractions: multiply and read down (section 4.4)

Watch 0.375 end, and 0.3 go on for ever.

```python
from fractions import Fraction

def fraction_to_base(f, b, places=8):
    f = Fraction(f)
    digits, seen = '', []
    while f and len(digits) < places:
        if f in seen:
            print('  this fraction came before: the digits repeat')
            break
        seen.append(f)
        p = f * b
        d = int(p)
        print(f'  {float(f):g} × {b} = {float(p):g} → {d}')
        digits += '0123456789ABCDEF'[d]
        f = p - d
    print('Read down: 0.' + digits)

fraction_to_base('0.375', 2)
fraction_to_base('0.3', 2)
```

## 6. Characters and their codes

`ord()` gives a character's code (its Unicode code point; for English letters this is the ASCII code). `chr()` goes back.

```python
print(ord('A'), ord('a'), ord('0'), ord(' '))
print(chr(65), chr(97 + 12))
word = 'RAMAN'
print([ord(c) for c in word])
print([format(ord(c), '07b') for c in word])
```

## 7. Unicode: code points and bytes

`.encode('utf-8')` gives the bytes that are stored. Compare UTF-8 and UTF-32.

```python
for text in ['A', '₹', 'अ', '😀', 'हम सब एक']:
    u8 = text.encode('utf-8')
    u32 = text.encode('utf-32-be')
    print(f'{text!r:12} code points: {[hex(ord(c)) for c in text]}')
    print(f'{"":12} UTF-8: {len(u8)} bytes {u8.hex(" ").upper()}   UTF-32: {len(u32)} bytes')
```
