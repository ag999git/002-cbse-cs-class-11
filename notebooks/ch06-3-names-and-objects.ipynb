## 3. Dynamic typing

The same name, tied to objects of different types:

```python
marks = 92
print(type(marks))
marks = "A+"
print(type(marks))
marks = 9.2
print(type(marks))
```

## 4. Swapping

With a third variable (the empty third glass), and Python's way:

```python
x, y = 5, 7
t = x
x = y
y = t
print(x, y)

x, y = y, x
print(x, y)
```
