## 1. See the tokens of any line

Python's own `tokenize` module cuts a line into tokens. The small function below names each one as the book does (section 6.4). Run it, then change `line` and run again.

```python
import io, keyword, tokenize

def show_tokens(line):
    count = 0
    for t in tokenize.generate_tokens(io.StringIO(line + "\n").readline):
        if t.type == tokenize.NAME:
            kind = "keyword" if keyword.iskeyword(t.string) else "identifier"
        elif t.type in (tokenize.NUMBER, tokenize.STRING):
            kind = "literal"
        elif t.type == tokenize.OP:
            kind = "punctuator" if t.string in "()[]{},:.;@" else "operator"
        elif t.type == tokenize.COMMENT:
            print(f"{t.string:<12} comment: skipped, not a token")
            continue
        else:
            continue
        count += 1
        print(f"{t.string:<12} {kind}")
    print(count, "tokens")

line = 'if marks >= 33: print("Pass")   # result'
show_tokens(line)
```

## 3. Is this a valid name?

`str.isidentifier()` checks the rules for the characters; `keyword.iskeyword()` checks the reserved words. A valid name needs both.

```python
def valid_name(name):
    return name.isidentifier() and not keyword.iskeyword(name)

for name in ["marks", "2nd_rank", "my marks", "for", "For", "_total", "sum$", "नाम"]:
    print(f"{name!r:<12} {valid_name(name)}")
```
