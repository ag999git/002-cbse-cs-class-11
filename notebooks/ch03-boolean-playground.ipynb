## 1. The truth table of any expression
Change the expression between the quotes, then run the cell.

```python
expr = "A·B' + C"
e = bl.parse(expr)
display(Markdown(bl.table_md(e, 'F')))
```

## 2. The complement, step by step (De Morgan's laws)

```python
expr = "A + B·C"
steps = bl.complement_steps(bl.parse(expr))
print("F' =", steps[0])
for s in steps[1:]:
    print('   =', s)
```

## 3. Are two expressions equal?
Use it to check a simplification or a law: it tries every row.

```python
left, right = "A + A'·B", "A + B"
print('Equal in every row' if bl.equal(bl.parse(left), bl.parse(right)) else 'NOT equal: compare the truth tables above')
```

## 4. The dual

```python
expr = "A + 0"
print('Dual:', bl.show(bl.dual(bl.parse(expr))))
```

## 5. Draw the circuit

```python
expr = "A'·B + A·C'"
bl.draw(bl.parse(expr), 'circuit.png', 'Y')
display(Image('circuit.png', width=520))
```
